1.	Create a list of 5 integers. Perform and print the result of the following operations: append, extend, insert, remove, pop, clear, index, count, sort, and reverse.

```python
# Question 1 - List of 5 Integers and Operations on the list

a = [24, 50, 33, 55, 39]
print("Initial list:", a)

a.append(2) #Appends 2 at the end of the list
print("\nafter append(2):", a)

a.extend([6, 7]) #Appends 6,7 at the end of the list
print("\nafter extend([6, 7]):", a)

a.insert(1, 26) # Inserts 26 at index 1
print("\nafter insert(1, 26):", a)

a.remove(50) #Removes first value of 50
print("\nafter remove(50):", a)

popped_val = a.pop(2) # Removes number at index 2
print(f"\nafter pop(2) [removed {popped_val}]:", a)

idx = a.index(39)
print("\nresult of index(39):", idx)

cnt = a.count(26) # returns count of exact value 26
print("\nresult of count(26):", cnt)

a.sort(reverse=True) # sorts in descending order
print("\nafter sort(reverse=True):", a)

a.reverse() # Reverses the entire list
print("\nafter reverse():", a)

a.clear() # Removes all elements in the list
print("\nafter clear():", a)
```

7.	Given a number, check if it's positive, and if it is also even. If not positive, print if it's zero or negative.

```python
# Question 7 - Check +ve, even , -ve, zero

a = int(input("Enter a number: "))

if a > 0:
    print(f"{a} is positive")
    if a % 2 == 0:
        print(f"{a} is even")
    else:
        print(f"{a} is odd")
elif a == 0:
    print("The number is zero")
else:
    print(f"{a} is negative")
```

11.	Create a small DataFrame manually with 10 rows and columns: 'Name', 'Age', 'City', and 'Salary'. Then:
- Use .info(), .describe()
- Select 'Name' and 'City' columns
- Drop 'City' column
- Fill any missing values in 'Salary' column with the mean
- Remove any duplicate rows


```python
# Question 11

import numpy as np
import pandas as pd

# 1. Create a DataFrame with 10 rows containing missing values and duplicates
data = { "Name": ["Alice","Bob", "Charlie", "David", "Eva", "Frank", "Grace", "Hank", "Alice", "Ivy", ],

         "Age": [25, 30, 35, 40, 28, 32, 45, 29, 25, 31],

         "City": ["New York","London","Paris","Tokyo","Berlin","London","Paris","Tokyo","New York","Berlin",],

         "Salary": [70000,80000,np.nan,95000,62000,80000,110000,np.nan,70000,68000,],
         }

df = pd.DataFrame(data)
print("\n---------------Initial Dataframe---------------")
print(df)
```

```python
# 2. USE .info()
print("\n---------------.info()---------------\n")
df.info()
```

```python
# 3. USE .describe()
print("\n---------------.describe()---------------\n")
df.describe()
```

```python
# 4. Select Name and City columns
df[["Name","City"]]
```

```python
# 5. Delete City Column
df = df.drop(columns="City", axis = 1)
df
```

```python
# 6. Fill missing values in 'Salary' column with mean
df["Salary"] = df["Salary"].fillna(df["Salary"].mean())
df
```

```python
# 7. Remove Duplicates
df = df.drop_duplicates()
df
```
