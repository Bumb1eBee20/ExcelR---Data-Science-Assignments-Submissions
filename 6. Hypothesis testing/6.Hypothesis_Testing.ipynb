# HYPOTHESIS TESTING

**Background**

Bombay hospitality Ltd. operates a franchise model for producing exotic Norwegian dinners throughout New England. The operating cost for a franchise in a week (W) is given by the equation W = $1,000 + $5X, where X represents the number of units produced in a week. Recent feedback from restaurant owners suggests that this cost model may no longer be accurate, as their observed weekly operating costs are higher.

**Objective :**

To investigate the restaurant owners' claim about the increase in weekly operating costs using hypothesis testing.

**Data Provided :**

●	The theoretical weekly operating cost model: W = $1,000 + $5X

●	Sample of 25 restaurants with a mean weekly cost of Rs. 3,050

●	Number of units produced in a week (X) follows a normal distribution with a mean (μ) of 600 units and a standard deviation (σ) of 25 units

**Assignment Tasks :**

## **Understand the problem**

W = 1000 + 5X

Given,

- Sample size : n = 25
- Sample mean cost : ˉxˉ = 3050
- Mean production : μX = 600
- Standard deviation : σX = 25
- Cost per unit : 5

Cost standard deviation as :

    σ = 5 x 25 = 125


**Theoretical mean cost**

put X = 600 into cost model:

    μW = 1000 + 5(600)
    μW = 4000

So we're comparing :

    Theoretical Mean = Rs. 4,000

    Against

    Sample Mean = Rs. 3,050

## **1. State the Hypothesis statement :**

    Null Hypothesis (H₀) : μ = 4000

    Alternative Hypothesis (H₁) : μ > 4000

This is a right-tailed test.

## **2. Calculate the Test Statistic :**

use the following formula to calculate the test statistic (z):

where:
  
    ●	ˉxˉ = sample mean weekly cost (Rs. 3,050)
    ●	μ = theoretical mean weekly cost according to the cost model (W = $1,000 + $5X for X = 600 units)
    ●	σ = 5*25 units
    ●	n = sample size (25 restaurants)

The formula is :
    
    Z = (ˉxˉ - μ)/(σ/sqrt(n))

Substitute the values :

    Z = (3050 - 4000)/(125/sqrt(25))

Since :

    sqrt(25) = 5

    Z = -950/25

    Z = -38

That's an extremely negative Z-score.


**Observation:**

 The sample mean weekly operating cost (₹3,050) is substantially lower than the theoretical mean cost (₹4,000). Therefore, the calculated Z-statistic is -38, which does not support the restaurant owners' claim that the actual operating cost is higher than the theoretical model.


## **3. Determine the probability and compare :**

Using the alpha level of 5% (alpha = 0.05)

The assignment specifies :

    α = 0.05

For a right-tailed Z-TEST, the critical value is :

    Z = 1.645

Our Result :

    -38 < 1.645

## **PYTHON IMPLEMENTATION**

```python
# ============================================
# HYPOTHESIS TESTING
# One-Sample Right-Tailed Z-Test
# ============================================

import math
from scipy.stats import norm

# --------------------------------------------
# Step 1: Define the given values
# --------------------------------------------

sample_mean = 3050       # Sample mean weekly operating cost
n = 25                   # Sample size

mean_units = 600         # Mean number of units produced
std_units = 25           # Standard deviation of units produced

fixed_cost = 1000        # Fixed weekly operating cost
cost_per_unit = 5        # Cost per unit

alpha = 0.05             # Significance level


# --------------------------------------------
# Step 2: Calculate theoretical mean cost
# --------------------------------------------

theoretical_mean = fixed_cost + (cost_per_unit * mean_units)

print("Theoretical Mean Weekly Cost:", theoretical_mean)


# --------------------------------------------
# Step 3: Calculate standard deviation of cost
# --------------------------------------------

sigma = cost_per_unit * std_units

print("Standard Deviation of Cost:", sigma)


# --------------------------------------------
# Step 4: State the hypotheses
# --------------------------------------------

print("\nHypotheses:")
print("H0: μ = 4000")
print("H1: μ > 4000")


# --------------------------------------------
# Step 5: Calculate the Z-test statistic
# --------------------------------------------

z_statistic = (
    (sample_mean - theoretical_mean)
    / (sigma / math.sqrt(n))
)

print("\nZ-test Statistic:", z_statistic)


# --------------------------------------------
# Step 6: Calculate the critical value
# --------------------------------------------

critical_value = norm.ppf(1 - alpha)

print("Critical Value:", critical_value)


# --------------------------------------------
# Step 7: Calculate the p-value
# --------------------------------------------

p_value = 1 - norm.cdf(z_statistic)

print("P-value:", p_value)


# --------------------------------------------
# Step 8: Make the decision
# --------------------------------------------

if z_statistic > critical_value:
    print("\nDecision: Reject the Null Hypothesis")
else:
    print("\nDecision: Fail to Reject the Null Hypothesis")


# --------------------------------------------
# Step 9: Final conclusion
# --------------------------------------------

if z_statistic > critical_value:
    print(
        "Conclusion: There is sufficient evidence to support "
        "the restaurant owners' claim that weekly operating "
        "costs are higher than the theoretical model suggests."
    )
else:
    print(
        "Conclusion: There is insufficient evidence to support "
        "the restaurant owners' claim that weekly operating "
        "costs are higher than the theoretical model suggests."
    )
```

## **4. Make a Decision :**

Compare the test statistic with the critical value to decide whether to reject the null hypothesis.

**Decision rule :**

- If Z > 1.645 -> Reject H₀
- If Z <= 1.645 -> Fail to reject H₀

**Our result :**    
    
    -38 < 1.645


**Therefore Fail to Reject the Null Hypothesis (H₀).**

As there is insufficient evidence to support the restaurant owners' claim that weekly operating costs are higher than the theoretical model suggests.

## **5. Conclusion :**

Based on the decision in the step 4, conclude whether there is strong evidence to support the restaurant owner's claim that the weekly operating costs are higher than the model suggests.

Since the calculated Z-statistic (-38) is less than the critical value (1.645), we fail to reject the null hypothesis. Therefore, at the 5% significance level, there is insufficient evidence to support the restaurant owners' claim that the weekly operating costs are higher than the theoretical model suggests.

# ESTIMATION AND CONFIDENCE INTERVALS

**Background**

In quality control processes, especially when dealing with high-value items, destructive sampling is a necessary but costly method to ensure product quality. The test to determine whether an item meets the quality standards destroys the item, leading to the requirement of small sample sizes due to cost constraints.

**Scenario**

A manufacturer of print-heads for personal computers is interested in estimating the mean durability of their print-heads in terms of the number of characters printed before failure. To assess this, the manufacturer conducts a study on a small sample of print-heads due to the destructive nature of the testing process.

**Data**

A total of 15 print-heads were randomly selected and tested until failure. The durability of each print-head (in millions of characters) was recorded as follows:

    1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32, 0.85, 1.07, 1.48, 1.20, 1.33, 1.18, 1.22, 1.29


**Assignment Tasks**

## **a. Build 99% Confidence Interval Using Sample Standard Deviation**

Assuming the sample is representative of the population, construct a 99% confidence interval for the mean number of characters printed before the print-head fails using the sample standard deviation. Explain the steps you take and the rationale behind using the t-distribution for this task.

Given,

    -> n = 15
    -> Population standard deviation is unknown
    -> sample standard deviation 's' is calculated

We use the t-distribution because the population standard deviation is unknown and the sample size is small.

**Formula :**
$$
CI = \bar{x} \pm t_{\alpha/2,df}\frac{s}{\sqrt{n}}
$$
    
Where:

- $\bar{x}$ = sample mean
- $s$ = sample standard deviation
- $n$ = sample size
- $df = n-1$ = degrees of freedom
- $t_{\alpha/2,df}$ = critical t-value

For a 99% confidence level:

$$
\alpha = 1 - 0.99 = 0.01
$$

$$
df = 15 - 1 = 14
$$

**PYTHON IMPLEMENTATION**

```python
import numpy as np
from scipy.stats import t, norm

# Print-head durability data (in millions of characters)
data = np.array([
    1.13, 1.55, 1.43, 0.92, 1.25,
    1.36, 1.32, 0.85, 1.07, 1.48,
    1.20, 1.33, 1.18, 1.22, 1.29
])

# Display the data
print("Print-head durability data:")
print(data)

# Calculate sample statistics
n = len(data)
sample_mean = np.mean(data)
sample_std = np.std(data, ddof=1)

print("Sample Size (n):", n)
print("Sample Mean:", round(sample_mean, 4))
print("Sample Standard Deviation:", round(sample_std, 4))

# Confidence level and significance level
confidence_level = 0.99
alpha = 1 - confidence_level

# Degrees of freedom
df = n - 1

# Critical t-value for a two-tailed 99% confidence interval
t_critical = t.ppf(1 - alpha / 2, df)

# Margin of error
margin_error_t = t_critical * (sample_std / np.sqrt(n))

# Confidence interval
lower_t = sample_mean - margin_error_t
upper_t = sample_mean + margin_error_t

print("Confidence Level:", confidence_level * 100, "%")
print("Alpha:", alpha)
print("Degrees of Freedom:", df)
print("Critical t-value:", round(t_critical, 4))
print("Margin of Error:", round(margin_error_t, 4))

print("\n99% Confidence Interval using t-distribution:")
print("Lower Bound:", round(lower_t, 4))
print("Upper Bound:", round(upper_t, 4))
```

### Interpretation of Part A

Since the population standard deviation is unknown and the sample size is small (n = 15), the sample standard deviation is used to estimate the population standard deviation. Therefore, the t-distribution is appropriate for constructing the confidence interval.

The **99% confidence interval** for the population mean durability, using the sample standard deviation and t-distribution, is approximately:

$$
\boxed{(1.104,\ 1.401)}
$$

This means that we are **99% confident that the true mean durability of the print-heads lies between approximately 1.104 and 1.401 million characters**.

## **b. Build 99% Confidence Interval Using Known Population Standard Deviation**

If it were known that the population standard deviation is 0.2 million characters, construct a 99% confidence interval for the mean number of characters printed before failure.

In this case, the population standard deviation is known:

$$
\sigma = 0.2
$$

Since the population standard deviation is known, the **standard normal (z) distribution** is used.

### Formula

$$
CI = \bar{x} \pm Z_{\alpha/2}\frac{\sigma}{\sqrt{n}}
$$

For a 99% confidence level:

$$
Z_{\alpha/2} \approx 2.576
$$

```python
# Known population standard deviation
population_std = 0.2

# Critical z-value for a two-tailed 99% confidence interval
z_critical = norm.ppf(1 - alpha / 2)

# Margin of error
margin_error_z = z_critical * (population_std / np.sqrt(n))

# Confidence interval
lower_z = sample_mean - margin_error_z
upper_z = sample_mean + margin_error_z

print("Known Population Standard Deviation:", population_std)
print("Confidence Level:", confidence_level * 100, "%")
print("Critical z-value:", round(z_critical, 4))
print("Margin of Error:", round(margin_error_z, 4))

print("\n99% Confidence Interval using z-distribution:")
print("Lower Bound:", round(lower_z, 4))
print("Upper Bound:", round(upper_z, 4))
```

### Interpretation of Part B

Since the population standard deviation is known to be 0.2 million characters, the standard normal (z) distribution is used to construct the confidence interval.

The **99% confidence interval** for the population mean durability, using the known population standard deviation and z-distribution, is approximately:

$$
\boxed{(1.119,\ 1.386)}
$$

This means that we are **99% confident that the true mean durability of the print-heads lies between approximately 1.119 and 1.386 million characters**.

## Comparison of the Two Confidence Intervals

| Method | Distribution | Standard Deviation Used | 99% Confidence Interval |
|---|---|---|---|
| Part A | t-distribution | Sample SD | (1.104, 1.401) |
| Part B | z-distribution | Population SD = 0.2 | (1.119, 1.386) |

The t-based interval is slightly wider because the population standard deviation is unknown and the t-distribution accounts for the additional uncertainty involved in estimating the standard deviation from the sample.

## Conclusion

The analysis provides two 99% confidence intervals for the mean durability of the print-heads.

- When the population standard deviation is **unknown**, the t-distribution gives a confidence interval of approximately **1.104 to 1.401 million characters**.
- When the population standard deviation is **known to be 0.2 million characters**, the z-distribution gives a confidence interval of approximately **1.119 to 1.386 million characters**.

Therefore, the estimated population mean durability falls within these respective ranges at the **99% confidence level**.
