```python
import pandas as pd
import numpy as np
```

```python
# 1.Load the patient dataset

# Raw URLs for direct CSV reading
patient_url = "https://raw.githubusercontent.com/Bumb1eBee20/ExcelR---Data-Science-Assignments-Submissions/22e13b7df95bb5fd6daedc8be3acb7cd7501135a/4.%20Basic%20Stats%20-%202/data/Patient_Data.csv"
billing_url = "https://raw.githubusercontent.com/Bumb1eBee20/ExcelR---Data-Science-Assignments-Submissions/22e13b7df95bb5fd6daedc8be3acb7cd7501135a/4.%20Basic%20Stats%20-%202/data/Billing_Data.csv"

df_patient = pd.read_csv(patient_url)
df_billing = pd.read_csv(billing_url)
```

```python
print("=" * 80)
print("PATIENT DATA")
print("=" * 80)

display(df_patient)
```

**Load Patient Dataset and Show Summary with info()**

```python
print("=" * 80)
print("PATIENT DATA SUMMARY")
print("=" * 80)

# Dataset information
df_patient.info()

# Shape
print("\nShape:", df_patient.shape)

# Number of rows
print("Number of Rows:", df_patient.shape[0])

# Number of columns
print("Number of Columns:", df_patient.shape[1])

# Null values
print("\nNull Values:")
print(df_patient.isnull().sum())

# Duplicate rows
print("\nDuplicate Rows:")
print(df_patient.duplicated().sum())
```

**Select Relevant Billing Columns**

```python
#Selecting Relevant Billing Columns
billing_patient = df_patient[
    ["PatientID", "Department", "Doctor", "BillAmount"]
]

print("=" * 80)
print("BILLING-RELATED PATIENT DATA")
print("=" * 80)

display(billing_patient)
```

**Drop Administrative Columns**

```python
# Dropping Administrative cols
df_patient = df_patient.drop(
    columns=["ReceptionistID", "CheckInTime"]
)

print("=" * 80)
print("PATIENT DATA AFTER DROPPING ADMINISTRATIVE COLUMNS")
print("=" * 80)

display(df_patient)
```

**Remove Duplicate Patient Records based on PatientID**

```python
print("Duplicate PatientIDs before removal:",
      df_patient["PatientID"].duplicated().sum())

df_patient = df_patient.drop_duplicates(
    subset="PatientID",
    keep="first"
).reset_index(drop=True)

print("Duplicate PatientIDs after removal:",
      df_patient["PatientID"].duplicated().sum())

display(df_patient)
```

**Fill Missing BillAmount with Mean**

```python
#Calculate Mean
mean_bill = df_patient["BillAmount"].mean()

print("Mean Bill Amount:", mean_bill)

df_patient["BillAmount"] = df_patient["BillAmount"].fillna(
    mean_bill
)

print("=" * 80)
print("PATIENT DATA AFTER FILLING MISSING BILL AMOUNTS")
print("=" * 80)

display(df_patient)
```

**Total Bill Amount Per Department**

```python

department_bill = (
    df_patient
    .groupby("Department")["BillAmount"]
    .sum()
    .reset_index()
)

department_bill.columns = [
    "Department",
    "Total_BillAmount"
]

print("=" * 80)
print("TOTAL BILL AMOUNT PER DEPARTMENT")
print("=" * 80)

display(department_bill)
```

```python
print("Missing BillAmount values:",
      df_patient["BillAmount"].isnull().sum())
```

**Merge Billing Dataset with Patient Dataset**

```python
print("=" * 80)
print("BILLING DATA")
print("=" * 80)

display(df_billing)

# .info()
print("\nBilling Data Information:")
df_billing.info()

#Checking Null values
print("\nNull Values:")
print(df_billing.isnull().sum())

#Checking Duplicate values
print("\nDuplicate Rows:")
print(df_billing.duplicated().sum())
```

```python
df_merged = pd.merge(
    df_patient,
    df_billing,
    on="PatientID",
    how="left"
)

print("=" * 80)
print("MERGED PATIENT + BILLING DATA")
print("=" * 80)

display(df_merged)
```

```python
print("=" * 80)
print("MERGED DATASET SUMMARY")
print("=" * 80)

#shape
print("Shape:", df_merged.shape)

# checking null values
print("\nNull Values:")
print(df_merged.isnull().sum())

# Checking Data types
print("\nData Types:")
print(df_merged.dtypes)
```

**Concatenate New Patients Row-wise**

```python
new_patients = pd.DataFrame({
    "PatientID": [106, 107],
    "Name": ["Frank", "Grace"],
    "Department": ["Cardiology", "Neurology"],
    "Doctor": ["Dr. Smith", "Dr. John"],
    "BillAmount": [6800.0, 5400.0],
    "ReceptionistID": [1, 2],
    "CheckInTime": [
        "2023-01-16 09:30",
        "2023-01-17 10:15"
    ]
})

print("=" * 80)
print("NEW PATIENTS")
print("=" * 80)

display(new_patients)
```

```python
# Select only columns matching the cleaned patient dataset
new_patients_clean = new_patients[
    ["PatientID", "Name", "Department", "Doctor", "BillAmount"]
]
new_patients_clean
```

```python
df_patient_updated = pd.concat(
    [
        df_patient,
        new_patients_clean
    ],
    axis=0,
    ignore_index=True
)

print("=" * 80)
print("PATIENT DATA AFTER ROW-WISE CONCATENATION")
print("=" * 80)

display(df_patient_updated)
```

```python
print("Original patient rows:", len(df_patient))
print("New patient rows:", len(new_patients_clean))
print("Updated patient rows:", len(df_patient_updated))
```

**Concatenate Billing Columns Column - wise**

```python
billing_columns = df_billing[
    ["InsuranceCovered", "FinalAmount"]
]

print("=" * 80)
print("BILLING COLUMNS")
print("=" * 80)

display(billing_columns)
```

```python
df_final = pd.concat(
    [
        df_patient,
        billing_columns
    ],
    axis=1
)

print("=" * 80)
print("FINAL DATASET AFTER COLUMN-WISE CONCATENATION")
print("=" * 80)

display(df_final)
```

**Final Cleaned Dataset**

```python
print("=" * 80)
print("FINAL CLEANED DATASET")
print("=" * 80)

display(df_final)

print("\nFinal Shape:", df_final.shape)

print("\nFinal Data Types:")
print(df_final.dtypes)

print("\nFinal Null Values:")
print(df_final.isnull().sum())
```
