# Exercise 1 : Prime Numbers

Write a Python program that checks whether a given number is prime or not. A prime number is a natural number greater than 1 that has no positive divisors other than 1 and itself.

```python
#Exercise 1 : Prime Numbers

import math
num = int(input("Enter a Number : "))
if num<=1:
  print(f"{num} is not a prime number")
else:
  for i in range(2,int(math.sqrt(num))+1):
    if num%i==0:
      print(f"{num} is not a prime number")
      break
  else:
      print(f"{num} is a prime number")
```

# Exercise 2 : Product of Random Numbers

Develop a Python program that generates two random numbers and asks the user to enter the product of these numbers. The program should then check if the user's answer is correct and display an appropriate message.

```python
# Exercise 2 : Product of Random Numbers

import random

num1 = random.randint(1,1000)
num2 = random.randint(1,1000)

product = num1*num2


while True:
  x = int(input(f"{num1} x {num2} = "))

  if (product == x):
    print(f"Correct! product of {num1} and {num2} is {x}")
    break

  else:
    print(f"Incorrect! product of {num1} and {num2} is not {x}")
    print("TRY AGAIN!!!\n")
```

# Exercise 4 : Problem Statement (question) : Hospital Billing System

A hospital charges its patients based on the following:

•	Room charges per day: ₹2,000

•	Doctor consultation fee (flat): ₹1,500

•	Lab test charges: Based on the number of tests taken. Each test costs ₹300

•	Medicine charges: Total cost of medicines provided

•	Discount: If the total bill (before discount) exceeds ₹10,000, a 10% discount is applied

Write a Python program to calculate the final bill for a patient who:

•	Stayed for 4 days

•	Had 3 lab tests

•	Had medicine charges of ₹2,400

```python
# Excercise 4 : Hospital Billing Systm

days = int(input("Enter NO.of Days Stayed : "))
lab_tests = int(input("Enter No. of Lab Tests : "))
Medical_charges = int(input("Enter Medicine Charges : "))
consultation_fee = 1500

total = ((days*2000)+(consultation_fee)+(lab_tests*300)+(Medical_charges))

if(total>10000):
  discount = total*0.1
  final_amount = total - discount
  print("\nCongratulations! 10% Discount is applied\n")
  print(f"Total Bill Amount is : {final_amount}")

else:
  final_amount = total
  print(f"Total Bill Amount is : {final_amount}")
```
